## Fstring (Format String)

```python
# fString
# it was introduced in python3.6

name = "chitti"
print("hello", name)
```

```python
#fstring example
name = "chitti"
print(f"hello {name}")
```

```python
name = "chitti"
age = 30
weight = 50

# Method 1: String Concatenation (+)
# Using + to combine multiple strings.
# str(age) converts integer into string because we cannot directly
# concatenate a string and an integer.

print("My name is " + name + ", I am " + str(age) + " years old")


# Method 2: Using comma (,)
# Comma separates multiple values inside print().
# Python automatically adds spaces between the values.
# No need to convert age into a string.

print("my name is", name, "I am", age, "years old")


# Method 3: Using f-string
# f-string was introduced in Python 3.6.
# We use f before the string.
# Variables are placed inside curly braces {}.
# Python automatically replaces variables with their values.

print(f"my name is {name} I am {age} years old")
```
