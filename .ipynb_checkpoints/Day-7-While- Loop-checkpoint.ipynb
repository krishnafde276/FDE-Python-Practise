# while loop

while loop is used to repeat block of code as long as given condition true

while loop can use index (i) to access list of elements

for loop is used to iterate over the sequence of elements (list, tuple, dictionary)

```python
#infinity loop

while True:
    print("hello")
```

```python

#infinity loop
while True:
    print("hello")
```

```python
#infinity loop print 1
i = 1
while True:
    print(i)
    
```

```python


#infinity loop print sequence order 0,1,2 .......
i = 0
while True:
    print(i)
    i = i+1  #incremental operator
    
```

```python

i = 1
while i <= 10:
    print(i)
    i = i+1
```

```python
i = 1

while i in range(1,5):
    password = input("Enter your password: ")

    if password == "akki":
        print("allow to login")
        break
    else:
        print("not allow")

    i = i+1
    
```
