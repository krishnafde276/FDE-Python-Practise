## Left Shift & Right Shift

```python
x=15
x<<=1
print(x)
```

```python
## 1. Example: Left Shift by 1
x = 15
x <<= 1
print(x)

# Before binary: 1111
# After binary:  11110

# Left shift (<<): Moves bits to the left
# Output: 30
```

```python
## 2. Example: Left Shift by 2
x = 19       # Binary: 10011
x <<= 2      # Binary: 1001100
print(x)

# Output: 76
```

```python
# Right Shift
x = 15
x >>= 1
print(x)

# Before binary: 1111
# After binary:  0111

# Output: 7
```

```python
# Right Shift
x = 15
x >>= 1
print(x)

# Before binary: 1111
# After binary:  0111


# Output: 7
```

```python
# Right Shift
x = 15
x >>= 2
print(x)

# Before binary: 1111
# After binary:  0011

# Output: 3
```

```python
# Left Shift

15          # 1111
15 << 1     # 11110
15 << 2     # 111100
15 << 3     # 1111000
```

```python
# Right Shift

15          # 1111
15 >> 1     # 00111
15 >> 2     # 00011
15 >> 3     # 00001
```
