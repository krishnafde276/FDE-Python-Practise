# #Loops

for, while, break, nested for loop, for loop if combination, for loop different types, operators

Indentation and syntax follow the same process as if.


```python
# for loop we can iterate over sequence of items, list, tuple, range, dictionary etc
# for loop with String Example

a = "python"

for i in a:
    print(i)
```

```python
#for loop with list example

number = ["aws", "azure", "gcp"]

for i in number:
    print(i)
```

```python
#for loop with break

number = [1,2,3,4,5,6]

for i in number:
    if i == 2:
        break
    print(i)
```

```python
## for loop with dictionary

number = {
    "name": "veera",
    "course": "fde"
}

for i in number.items():
    print(i)
```

```python
# for Loop calling keys

number = {
    "name": "veera",
    "course": "fde"
}

for i in number.keys():
    print(i)
```

```python
number = [10,20,30,40,50]

for i in number:
    if i == 40:
        print("number found")
        break
    print(i)
```

```python
# For Loop – Range Example

total = 1
dev = range(1,5)

for i in dev:
    total = total+i

print("total", total)


# First iteration
# i = 1
# total = total+i
#       = 1+1
# total = 2


# Second iteration
# i = 2
# total = total+i
#       = 2+2
# total = 4


# Third iteration
# i = 3
# total = total+i
#       = 4+3
# total = 7


# 4th iteration
# i = 4
# total = total+i
#       = 7+4
# total = 11


# 5th iteration
# i = 5
# total = total+i
#       = 11
```

```python
x = [1,2,3,4,5,6,7]

for i in x:
    if i % 2 == 0:
        print(i)
    else:
        print("odd number")
```

```python
number = [10, -5, 0, 20, -8]

for num in number:
    if num > 0:
        print(num, "is positive")
    elif num < 0:
        print(num, "is negative")
    else:
        print(num, "is zero")
```

x                       Variable
=                       Assignment operator
[1,2,3,4,5,6,7]         List
for                     Keyword
i                       Variable
in                      Iteration keyword
if                      Conditional keyword
%                       Arithmetic operator (Modulus)
==                      Comparison operator
print()                 Built-in function
else                    Conditional keyword
Indentation             Spaces at the beginning of a code line
:                       Colon (used to start a code block)
