```python

## String Immutability

### String cannot be modified once we declared

# String Immutable means you cannot modify the existing string after it is created
```

```python
word = "PYTHON"

word = "J" + word[1:]
print(word)
```

```python

word = "PYTHON"
new_word = "J" + word[1:]

print("old:", word)
print("new:", new_word)
```

```python
#join
words = ["python", "is", "easy"]
sentence= " " .join(words)
print(words)
print(sentence)
```

```python

## List

## characteristics: index, slicing, mutable
```

```python

fruits = ["apple", "banana", "mango", "orange", "grapes"]

print(fruits)
print(fruits[4])  # calling position by using index

# list with slicing
print(fruits[1:4])
```

```python

# append: by default adds a new element to end of the list

fruits = ["apple", "banana", "orange"]
fruits.append("mango")  # function method use parentheses ()

print(fruits)
```

```python

fruits = ["apple", "banana", "orange"]
fruits.append(["mango", "cherry"])  # python adds the entire list as one single element

print(fruits)
```

```python

fruits = ["apple", "banana", "orange"]
fruits.append(["mango", "cherry"])  # python adds the entire list as one single element

print(fruits)
```

```python

# insert method add element into specific position

fruits = ["apple", "banana", "orange"]
fruits.insert(1, "mango")

print(fruits)
```

```python

# example-2
# insert method add element into specific position

fruits = ["apple", "banana", "orange"]
fruits.insert(1, "grape")

print(fruits)
```

```python

# we can use del method to delete specific element by calling index
# del method not support by element value

fruits = ["apple", "banana", "orange"]
del fruits[1]

print(fruits)
```

```python

# pop method Remove last element

fruits = ["apple", "banana", "orange"]
fruits.pop()

print(fruits)
```

```python

# pop method Remove using index specific element we can remove

fruits = ["apple", "banana", "orange"]
fruits.pop(1)

print(fruits)
```

```python

# pop method Remove using index specific element we can remove

fruits = ["apple", "banana", "orange"]
fruits.pop(1)

print(fruits)
```

```python

## tuple

### indexing, slicing

##### Please follow list examples for tuple
```

```python

fruits = ("apple", "banana", "orange")
print(fruits)

fruits = ("apple", "banana", "orange")
print(fruits[1])

fruits = ("apple", "banana", "orange")
print(fruits[2:3])

```

```python

## tuple Immutable

### These methods not support - add, update, remove, insert, extend, sort
```

```python

## Sets

### use { }
### Store unique values
### not allow duplicates
```

```python

numbers = {10, 20, 30, 40, 50}
print(numbers)
```

```python

# the set output order may vary, no guarantee for order

numbers = {10, 20, 10, 30, 40, 50}
print(numbers)

# duplicate values automatically removed in sets
```

```python

## Set mutable -> update, add, remove, insert etc ...
```

```python

numbers = {10, 20, 30, 40, 50}
value = numbers.pop()

print(value)
```

```python

student = {
    "name": "veera",
    "cousre": "FDE",
    "age": 30
}

print(student)
```

```python

# if i declare duplicate values allowed last value

student = {
    "name": "veera",
    "cousre": "FDE",
    "age": 30,
    "age": 40
}

print(student)
```

```python

student = {
    "name": "veera",
    "cousre": "FDE",
    "age": 30
}

student["age"] = 40

print(student)
```

```python
## Adding a new record
student = {
    "name": "veera",
    "cousre": "FDE",
    "age": 30
}

student["city"] = "hyd"

print(student)
```

```python
## Renaming a key using pop()
student = {
    "name": "veera",
    "cousre": "FDE",
    "age": 30
}

student["student_name"] = student.pop("name")

print(student)
```
