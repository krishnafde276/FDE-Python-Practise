Comparison Operators ==, !=a>b, a<b; a<=b,a<=b


```python
## 1. Equal to (==)
a = 5
b = 10

print(a == b)
 
# Output: False
```

```python
##2. Not Equal to (!=)

a = 5
b = 10

print(a != b)

# Output: True
```

```python
## 3. Greater Than (>)
a = 5
b = 10

print(a != b)

# Output: True
```

```python
# Example

age = 18

if age >= 18:
    print("you are eligible to vote.")
else:
    print("you are not eligible to vote.")

# Output:
# you are eligible to vote.
```
