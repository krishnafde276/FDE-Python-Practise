## Python Conditions if, elif, else, Nested condition

### Rule-1: use colon (:) every if, else, elif
### Rule-2: Indentation mandatory
### Rule-3: elif, else cannot stand alone
### correct indentation example


```python
#simple if
age = 20
if age >= 18:
    print("You are eleigible to vote")


```

```python
#simple if and else
age = 17
if age >= 18:
    print("You are eleigible to vote")
else:
    print("You are not eligible to vote")
```

```python
#pass value dynamically using input()
#by default input function consider string

age = int(input("enter age"))
if age >= 18:
    print("You are eleigible to vote")
else:
    print("You are not eligible to vote")
```

```python
## login creds validations

password = "Akki123"
if password == "Akki123":
    print("login succesfull")
else:
    print("Invalid password")
```

```python
# if elif

marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("fail")
```

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("fail")

# Each condition is checked independently
# It is executing multiple matching blocks
```

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("fail")

# Execute only one matching block
```
